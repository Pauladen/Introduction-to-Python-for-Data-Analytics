**Create dataframes of category and numeric columns**

```python
df_num = df_key.select_dtypes(include=['float', 'int'])
df_num.columns
```

```python
df_cat = df_key.select_dtypes(include=['object'])
df_cat.columns
```

**Import SimpleImputer fro sklearn**

```python
from sklearn.impute import SimpleImputer
```

**Handling dataframes of numeric features**

```python
# Instantiate SimpleImputer
si_num = SimpleImputer(strategy='median')
df_num_imputed = si_num.fit_transform(df_num)
df_num_imputed
```

**Convert si_num back to dataframe**

```python
df_num_df = pd.DataFrame(df_num_imputed, columns= df_num.columns, index=df_num.index)
df_num_df.head()
```

**Confirm that there is no missing numeric values**

```python
df_num_df.isnull().sum()
```

**Handling dataframes of category features**

```python
# Instantiate SimpleImputer
si_cat = SimpleImputer(strategy='most_frequent')
df_cat_imputed = si_cat.fit_transform(df_cat)
df_cat_imputed
```

**Convert si_num back to dataframe**

```python
df_cat_df = pd.DataFrame(df_cat_imputed, columns= df_cat.columns, index=df_cat.index)
df_cat_df.head()
```

**Confirm that there is no missing numeric values**

```python
df_cat_df.isnull().sum()
```

**Concat the imputed numeric and category dataframes to have a new dataframe**

```python
df_eda = pd.concat([df_num_df, df_cat_df], axis=1)
df_eda.columns
```

```python
# melt dataframe
df_melted = pd.melt(df_num_df)
sns.set_style('darkgrid')
plt.figure(figsize=(15, 10))
sns.boxplot(data=df_melted, x='variable', y='value', color='skyblue')
plt.xticks(rotation=90)
plt.xlabel('Features')
plt.ylabel('Numerical Values')
plt.title('Box Plots Checks for Outliers In Numeric Features')
plt.tight_layout()
plt.show()
```

**There are so much outliers in the:**

* CURRENT_LAND_VALUE
* CURRENT_IMPROVEMENT_VALUE

**We will handle these outliers using Interquantile Range Method**

**It seams ~, <, >, and | are functions and should be followed by parenthesis**

```python
df_eda.duplicated().sum()
```

```python
df_eda.drop_duplicates(inplace=True)
```

```python
df_eda.duplicated().sum()
```

```python
df_nume = df_eda.select_dtypes(include=['float', 'int'])
df_nume.head()
```

```python
df_cate = df_eda.select_dtypes(include='object')
df_cate.head()
```

```python
Q1 = df_nume.quantile(0.25)
Q3 = df_nume.quantile(0.75)
IQR = Q3 - Q1
df_nume_inliers = df_nume[~((df_nume < (Q1 - 1.5 * IQR))|(df_nume > (Q3 + 1.5 * IQR))).any(axis=1)]
df_nume_inliers.head()
```

**Allign the index of df_nume_inliers with that of df_cate:**

```python
df_cate = pd.DataFrame(df_cate, columns=df_cate.columns, index=df_nume_inliers.index)
df_cate.head()
```

```python
print(df_nume_inliers.shape)
print(df_cate.shape)
```

```python
df_clean = pd.concat([df_nume_inliers, df_cate], axis=1)
df_clean.head()
```

```python
df_clean.describe()
```

```python
# melt dataframe
df_melted = pd.melt(df_nume_inliers)
sns.set_style('darkgrid')
plt.figure(figsize=(15, 10))
sns.boxplot(data=df_melted, x='variable', y='value', color='skyblue')
plt.xticks(rotation=90)
plt.xlabel('Features')
plt.ylabel('Numerical Values')
plt.title('Box Plots of Numeric Features after Handling Outliers')
plt.tight_layout()
plt.show()
```

# Average of Tax Levy By Average Land Value over Zones

```python
Avg_Tax_zone = df_clean.groupby('ZONING_CLASSIFICATION')['TAX_LEVY'].mean()
Avg_Land_Val_zone = df_clean.groupby('ZONING_CLASSIFICATION')['CURRENT_LAND_VALUE'].mean()
plt.figure(figsize=(15, 15))
plt.scatter(Avg_Tax_zone, Avg_Land_Val_zone, marker='o', alpha=0.7, color='darkgreen')
plt.title('Average of Tax Levy By Average Land Value over Zones')
plt.xlabel('Average Tax Levy ($)')
plt.ylabel('Average Land Value ($)')
plt.show()
```

**Better approach:**

* **Create DataFrame (allows for sorting)**

* **Annotate each point**

ha means `horizontal alignment`

```python
# Group and calculate averages
Avg_Tax_zone = df_clean.groupby('ZONING_CLASSIFICATION')['TAX_LEVY'].mean()
Avg_Land_Val_zone = df_clean.groupby('ZONING_CLASSIFICATION')['CURRENT_LAND_VALUE'].mean()

## Combine into one DataFrame for easier plotting (Allows for sorting)
zone_data = pd.DataFrame({
    'Avg_Tax_Levy': Avg_Tax_zone,
    'Avg_Land_Value': Avg_Land_Val_zone
}).sort_values(by='Avg_Tax_Levy')  # Sort for cleaner visual layout

# Plot
plt.figure(figsize=(12, 8))
plt.scatter(zone_data['Avg_Tax_Levy'], zone_data['Avg_Land_Value'], 
            marker='o', alpha=0.8, color='darkgreen')

## Annotate each point
for idx, row in zone_data.iterrows():
    plt.annotate(idx, 
                 (row['Avg_Tax_Levy'], row['Avg_Land_Value']),
                 textcoords="offset points", 
                 xytext=(5,5), ha='center', fontsize=9, color='red')
    
# Labels and grid
plt.title('Average Tax Levy vs. Average Land Value by Zoning Classification')
plt.xlabel('Average Tax Levy ($)')
plt.ylabel('Average Land Value ($)')
plt.tight_layout()
plt.show()
```

# Average of Tax Levy By Average Land Value over LEGAL TYPE

```python
Avg_Tax_legal = df_clean.groupby('LEGAL_TYPE')['TAX_LEVY'].mean()
Avg_Land_Val_legal = df_clean.groupby('LEGAL_TYPE')['CURRENT_LAND_VALUE'].mean()
# Create dataframe, and sort
legal_df = pd.DataFrame({
    'Avg_Tax_legal': Avg_Tax_legal, 
    'Avg_Land_Val_legal': Avg_Land_Val_legal
}).sort_values(by='Avg_Tax_legal')
## Plot
plt.figure(figsize=(12, 8))
plt.scatter(legal_df['Avg_Tax_legal'], legal_df['Avg_Land_Val_legal'], color='red', marker='o', alpha=0.8)
# Annotate
for idx, row in legal_df.iterrows():
    plt.annotate(idx, (row['Avg_Tax_legal'], row['Avg_Land_Val_legal']), textcoords='offset points', xytext=(5, 5), ha='center', fontsize=8, color='blue')
plt.title('Average of Tax Levy By Average Land Value over LEGAL TYPE')
plt.xlabel('Average Tax ($)')
plt.ylabel('Average Land Value ($)')
plt.tight_layout()
plt.show()
```
